```python
# training times

import matplotlib.pyplot as plt
from datetime import datetime

plt.rcParams['font.serif'] = ['Times New Roman']

# 定义训练时间数据
training_data = [
    ("PRICE", "2025-05-20 16:16:47", "2025-05-21 20:13:12"),
    ("Simple PRICE", "2025-05-15 19:08:23", "2025-05-15 20:56:18"),
    ("Baseball PRICE", "2025-05-22 11:32:53", "2025-05-22 13:08:15")
]

# 计算训练时长（以秒为单位）
durations = []
for _, start_time, end_time in training_data:
    start = datetime.strptime(start_time, "%Y-%m-%d %H:%M:%S")
    end = datetime.strptime(end_time, "%Y-%m-%d %H:%M:%S")
    duration = (end - start).total_seconds()
    durations.append(duration)

# 将时长转换为分钟
durations = [d / 60 for d in durations]

# 提取训练类型
training_types = [data[0] for data in training_data]

# 设置图片清晰度
plt.rcParams['figure.dpi'] = 300

# 设置 matplotlib 支持中文
plt.rcParams['font.sans-serif'] = ['WenQuanYi Zen Hei']
plt.rcParams['axes.unicode_minus'] = False

# 透明度列表
alphas = [0.4, 0.6, 0.8]

# 创建更宽的图形，增加高度方向的空间
plt.figure(figsize=(10, 4))  # 调整宽度和高度

# 绘制横向柱状图，使用浅蓝色并设置不同透明度
for i in range(len(training_types)):
    plt.barh(
        training_types[len(training_types) - 1 - i], 
        durations[len(durations) - 1 - i],
        color=(0.68, 0.85, 0.90), 
        alpha=alphas[i], 
        height=0.4  # 增加高度，减少紧凑感
    )

# 设置 x 轴标签
plt.xlabel('training times')
# 去掉 y 轴标签
plt.ylabel('')

# 添加数据标签
for i, v in enumerate(durations):
    plt.text(v, len(durations) - 1 - i, f'{v:.2f}', ha='left', va='center')

# 调整子图参数，设置适当的垂直间距
plt.subplots_adjust(hspace=0.1)

plt.savefig('training_times.pdf', format='pdf', bbox_inches='tight')
# 显示图形
plt.show()
```

```python
# domain scores

import json
import os
import matplotlib.pyplot as plt

from rank_examples import show_ranks

plt.rcParams['font.serif'] = ['Times New Roman']

sorted_domain_scores = show_ranks()
selected_domains = [domain for domain in list(sorted_domain_scores.keys())[-5:]]

with open('results/domain_scores/hardest_counts_per_epoch.json', 'r', encoding='utf-8') as fin:
    hardest_counts = json.load(fin)
        
with open('results/domain_scores/easier_counts_per_epoch.json', 'r', encoding='utf-8') as fin:
    easier_counts = json.load(fin)
    
domain_scores = {}
for epoch in range(150):
    if epoch == 0:
        domain_scores[str(epoch)] = {domain_id: (hardest_counts[str(epoch)][domain_id] - easier_counts[str(epoch)][domain_id]) for domain_id in selected_domains}
    else:
        domain_scores[str(epoch)] = {domain_id: 
            (hardest_counts[str(epoch)][domain_id] - easier_counts[str(epoch)][domain_id]) * (1 - 0.5) + domain_scores[str(epoch-1)][domain_id] * 0.5
            for domain_id in selected_domains}
        
all_data = {}
file_names = {'25':'tubepricing', '16':'movielens', '2':'baseball', '6':'chembl', '21':'talkingdata'}
        
for domain_id in selected_domains:
    file_name = file_names[domain_id]

    x = []
    y = []

    for key in domain_scores.keys():
        x.append(int(key))
        y.append(domain_scores[str(key)][domain_id])
        
    all_data[file_name] = {
                'x': x,
                'y': y,
            }
    
plt.figure(figsize=(14, 8))

colors = plt.cm.tab10.colors 
line_styles = ['-', '--', '-.', ':']
markers = ['o', 's', '^', 'D', 'v', '<', '>', 'p', '*', 'h']

for i, (file_name, data_dict) in enumerate(all_data.items()):
    color = colors[i % len(colors)]
    line_style = line_styles[(i // len(colors)) % len(line_styles)]
    marker = markers[(i // (len(colors) * len(line_styles))) % len(markers)]
    marker_every = max(1, len(data_dict['x']) // 15)  
    
    plt.plot(data_dict['x'], data_dict['y'], 
            label=file_name, 
            color=color, 
            linestyle=line_style,
            marker=marker, 
            markersize=6,
            markevery=marker_every,
            linewidth=2)

plt.legend(loc='best', fontsize=10, framealpha=0.9, ncol=2)

plt.xlabel('epoch')
plt.ylabel('hardest_scores - easier_scores')
plt.grid(True)

plt.savefig('./score_evolution.pdf', format='pdf', bbox_inches='tight')
plt.show()

plt.close()
```

```python
# CEDaRe VS DoReMi VS Bootstrap Resampling qerror 90% and 99%

import matplotlib.pyplot as plt
import numpy as np

# 全局样式设置
plt.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Times New Roman', 'DejaVu Serif'],
    'font.size': 12,
    'axes.titlesize': 14,
    'axes.labelsize': 13,
    'xtick.labelsize': 11,
    'ytick.labelsize': 11,
    'legend.fontsize': 10,
    'figure.dpi': 300,
    'savefig.dpi': 300,
    'savefig.bbox': 'tight',
    'axes.edgecolor': '#333333',
    'axes.linewidth': 1.0,
    'grid.color': '#aaaaaa',
    'grid.alpha': 0.35,
    'grid.linestyle': '--',
    'grid.linewidth': 0.5,
})

# 数据
datasets = ['imdb', 'stats', 'ergastf1', 'genome']
models = ['CEDaRe', 'DoReMi', 'Bootstrap']

data = {
    'imdb': {
        'CEDaRe': [8.72, 44.02],
        'DoReMi': [5.32, 47.27],
        'Bootstrap': [7.71, 51.80],
    },
    'stats': {
        'CEDaRe': [25.98, 849.26],
        'DoReMi': [14.08, 1923.88],
        'Bootstrap': [16.80, 2044.02],
    },
    'ergastf1': {
        'CEDaRe': [8.63, 56.05],
        'DoReMi': [7.31, 37.82],
        'Bootstrap': [9.16, 94.63],
    },
    'genome': {
        'CEDaRe': [14.30, 72.78],
        'DoReMi': [7.50, 143.11],
        'Bootstrap': [4.90, 139.23],
    },
}

# 配色方案
model_colors = {
    'CEDaRe': '#4C72B0',      # 柔和蓝色
    'DoReMi': '#DD8452',      # 柔和橙色
    'Bootstrap': '#55A868',   # 柔和绿色
}

n_datasets = len(datasets)
n_models = len(models)

# 布局参数
model_group_width = 0.22       # 每个 model 组宽度
bar_width = 0.08               # 每根柱子宽度
bar_gap = 0.015                # 同一 model 两根柱子间隙
group_gap = 0.025              # model 组间间隙

fig, ax = plt.subplots(figsize=(10, 4.8))

from matplotlib.patches import Patch

dataset_positions = np.arange(n_datasets)

for d_idx, dataset in enumerate(datasets):
    # 计算当前 dataset 组起始 x
    total_width = n_models * model_group_width + (n_models - 1) * group_gap
    group_start = dataset_positions[d_idx] - total_width / 2

    for m_idx, model in enumerate(models):
        # 当前 model 组的中心
        model_x_start = group_start + m_idx * (model_group_width + group_gap)

        # 同一 model 内: 90% 在左, 99% 在右
        x_90 = model_x_start + (model_group_width / 2) - bar_width - bar_gap / 2
        x_99 = model_x_start + (model_group_width / 2) + bar_gap / 2

        val_90 = data[dataset][model][0]
        val_99 = data[dataset][model][1]

        # 90% 柱子（浅色）
        ax.bar(
            x_90, val_90, bar_width,
            color=model_colors[model],
            alpha=0.50,
            edgecolor='white',
            linewidth=0.4,
            zorder=3,
        )
        # 99% 柱子（深色 + 斜线填充）
        ax.bar(
            x_99, val_99, bar_width,
            color=model_colors[model],
            alpha=0.92,
            edgecolor='white',
            linewidth=0.3,
            hatch='///',
            zorder=3,
        )

        # 在每个 model 组下方标注 model 名称
        ax.text(
            model_x_start + model_group_width / 2,
            -0.12,
            model,
            ha='center', va='top',
            fontsize=7.5,
            color='#444444',
            transform=ax.get_xaxis_transform(),
        )

# 横轴刻度 - 放在每组柱子下方
ax.set_xticks(dataset_positions)
ax.set_xticklabels([d.capitalize() for d in datasets], fontweight='bold', fontsize=11)

# 将横轴刻度标签稍微下移，给 model 名称留出空间
ax.tick_params(axis='x', pad=8)

# 纵轴
ax.set_ylabel('Q-error', fontweight='bold')
ax.set_yscale('log')
ax.set_ylim(bottom=0.8, top=8000)

# 网格
ax.yaxis.grid(True, which='major')
ax.set_axisbelow(True)

# 左侧纵轴添加小刻度线
ax.tick_params(axis='y', which='both', length=4)

# 图例: 模型 + 百分位
legend_elements = [
    Patch(facecolor='#999999', alpha=0.5, edgecolor='white', linewidth=0.4, label='90%'),
    Patch(facecolor='#999999', alpha=0.92, edgecolor='white', linewidth=0.3, hatch='///', label='99%'),
]
for model in models:
    legend_elements.append(
        Patch(facecolor=model_colors[model], alpha=0.7, edgecolor=model_colors[model], label=model)
    )

legend = ax.legend(
    handles=legend_elements,
    loc='upper center',
    bbox_to_anchor=(0.5, 1.18),
    ncol=5,
    frameon=True,
    fancybox=True,
    edgecolor='#cccccc',
    fontsize=9.5,
    handlelength=1.2,
    handleheight=0.8,
    columnspacing=1.2,
    borderpad=0.4,
)

plt.tight_layout()
plt.savefig('cedare_doremi_bootstrap_qerror.pdf', format='pdf', bbox_inches='tight')
plt.savefig('cedare_doremi_bootstrap_qerror.png', format='png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Impact of eps and min_samples

import matplotlib.pyplot as plt
import numpy as np

# 全局样式设置
plt.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Times New Roman', 'DejaVu Serif'],
    'font.size': 11,
    'axes.titlesize': 12,
    'axes.labelsize': 11,
    'xtick.labelsize': 9,
    'ytick.labelsize': 9,
    'legend.fontsize': 9,
    'figure.dpi': 300,
    'savefig.dpi': 300,
    'savefig.bbox': 'tight',
    'axes.edgecolor': '#333333',
    'axes.linewidth': 0.8,
    'grid.color': '#bbbbbb',
    'grid.alpha': 0.3,
    'grid.linestyle': '--',
    'grid.linewidth': 0.4,
})

# 数据
datasets = ['imdb', 'stats', 'ergastf1', 'genome']
# 含义: eps (DBSCAN 邻域半径), min_samples (核心点最少样本数)
models = ['eps0.002_min150', 'eps0.0025_min200', 'eps0.0035_min200']
# 短标签，便于在图中显示
short_labels = ['ε=0.002\nmin=150', 'ε=0.0025\nmin=200', 'ε=0.0035\nmin=200']

data = {
    'imdb': {
        'eps0.002_min150': [1.61, 45.05],
        'eps0.0025_min200': [1.61, 40.83],
        'eps0.0035_min200': [1.62, 47.45],
    },
    'stats': {
        'eps0.002_min150': [2.06, 1494.49],
        'eps0.0025_min200': [2.05, 2206.25],
        'eps0.0035_min200': [2.11, 2322.27],
    },
    'ergastf1': {
        'eps0.002_min150': [1.59, 75.42],
        'eps0.0025_min200': [1.66, 55.90],
        'eps0.0035_min200': [1.68, 69.23],
    },
    'genome': {
        'eps0.002_min150': [1.84, 55.07],
        'eps0.0025_min200': [2.19, 34.81],
        'eps0.0035_min200': [1.83, 36.39],
    },
}

# 提取两个分位点的值
p50 = {ds: [data[ds][m][0] for m in models] for ds in datasets}
p99 = {ds: [data[ds][m][1] for m in models] for ds in datasets}

# 配色
color_50 = '#4C72B0'   # 50% 蓝色
color_99 = '#DD8452'   # 99% 橙色

# 绘图
fig, axes = plt.subplots(1, 4, figsize=(12, 4.2), sharey=True)
x = np.arange(len(models))
width = 0.30

for i, ds in enumerate(datasets):
    ax = axes[i]

    # 分组柱状图
    bars1 = ax.bar(
        x - width / 2, p50[ds], width,
        label='50%', color=color_50, alpha=0.75,
        edgecolor='white', linewidth=0.4, zorder=3,
    )
    bars2 = ax.bar(
        x + width / 2, p99[ds], width,
        label='99%', color=color_99, alpha=0.85,
        edgecolor='white', linewidth=0.4,
        hatch='///', zorder=3,
    )

    # 子图标题
    title = ds.capitalize() if ds != 'ergastf1' else 'ErgastF1'
    ax.set_title(title, fontweight='bold', pad=8)

    # X 轴
    ax.set_xticks(x)
    ax.set_xticklabels(short_labels, fontsize=7.5, linespacing=0.9)

    # 对数纵轴
    ax.set_yscale('log')
    ax.set_ylim(bottom=0.8, top=8000)

    # 网格
    ax.yaxis.grid(True, which='major')
    ax.set_axisbelow(True)

    # 只有第一个子图显示 Y 轴标签
    if i == 0:
        ax.set_ylabel('Q-error', fontweight='bold')
    else:
        ax.set_yticklabels([])

    # 数值标签（用指数偏移避免重叠）
    for bar in bars1:
        h = bar.get_height()
        offset = 1.15 if h < 100 else 1.08
        ax.text(
            bar.get_x() + bar.get_width() / 2., h * offset,
            f'{h:.2f}', ha='center', va='bottom',
            fontsize=6.5, color=color_50, fontweight='bold',
        )
    for bar in bars2:
        h = bar.get_height()
        offset = 1.15 if h < 100 else 1.08
        ax.text(
            bar.get_x() + bar.get_width() / 2., h * offset,
            f'{h:.2f}', ha='center', va='bottom',
            fontsize=6.5, color=color_99, fontweight='bold',
        )

# 统一图例（放在顶部中间）
fig.legend(
    ['50%', '99%'],
    loc='upper center',
    bbox_to_anchor=(0.5, 1.04),
    ncol=2,
    frameon=True,
    fancybox=True,
    edgecolor='#cccccc',
    fontsize=10,
    handlelength=1.5,
    handleheight=0.8,
    columnspacing=2.0,
)

plt.tight_layout()
plt.savefig('impact_eps_minsamples.pdf', format='pdf', bbox_inches='tight')
plt.savefig('impact_eps_minsamples.png', format='png', dpi=300, bbox_inches='tight')
plt.show()
```
